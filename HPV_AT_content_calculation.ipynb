```python
from Bio import SeqIO
import pandas as pd

# hpv_seq_file = open(r"C:\Users\thaka\OneDrive\Desktop\CRISPR-cas12a test\GCF_001274345.1_ViralProj293723_genomic.fna","rt") #The dataset we are using (find it in this repo only, it is subject to change)
hpv_seq_file = open(r"C:\Users\thaka\OneDrive\Desktop\CRISPR-cas12a test\gene.fna","rt") #The dataset we are using (find it in this repo only, it is subject to change)

sequences_obj = SeqIO.parse(hpv_seq_file, "fasta") # Use SEQIO to parse the fasta file into an object format
sequences = [seq for seq in sequences_obj] # data converted from object to list format for easier manipulation
sequences_df = pd.DataFrame([
    {"id": s.id, "seq": str(s.seq)} for s in sequences
]) # data converted from object to pandas dataframe format for easier manipulation

print(f"Starting Base pairs {sequences[0].seq[:50]}") # Prints the first 50 base pairs just to check if the data is being read correctly. You can change the index to see more or less base pairs.
```

```python
AT_content = sum(seq.seq.count('A') + seq.seq.count('T') for seq in sequences) / sum(len(seq.seq) for seq in sequences) #general formula to calculate AT content of a sequence. It counts the number of A's and T's in the sequence and divides it by the total length of the sequence.
print(f"AT content of HPV sequence : {AT_content * 100:.3f}%")
```
