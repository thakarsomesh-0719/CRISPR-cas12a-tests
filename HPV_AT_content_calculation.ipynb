```python
from Bio import SeqIO
import pandas as pd

seq_file = open(r"C:\Users\thaka\OneDrive\Desktop\CRISPR-cas12a test\GCF_001274345.1_ViralProj293723_genomic.fna","rt") #The dataset we are using (find it in this repo only, it is subject to change)

sequences_obj = SeqIO.parse(seq_file, "fasta") # Use SEQIO to parse the fasta file into an object format
sequences = [seq for seq in sequences_obj] # data converted from object to list format for easier manipulation
sequences_df = pd.DataFrame([
    {"id": s.id, "seq": str(s.seq)} for s in sequences
]) # data converted from object to pandas dataframe format for easier manipulation

print(f"Starting Base pairs {sequences[0].seq[:50]}") # Prints the first 50 base pairs just to check if the data is being read correctly. You can change the index to see more or less base pairs.
```

```python
AT_content = sum(seq.seq.count('A') + seq.seq.count('T') for seq in sequences) / sum(len(seq.seq) for seq in sequences) #general formula to calculate AT content of a sequence. It counts the number of A's and T's in the sequence and divides it by the total length of the sequence.
print(f"AT content of HPV sequence : {AT_content * 100:.3f}%")
```

```python
from sklearn.feature_extraction.text import CountVectorizer # I am still learning about this library, but it is used to convert text data into numerical data that can be used for machine learning models.

def to_kmers(seq, k=4):
    return ' '.join(seq[i:i+k] for i in range(len(seq) - k + 1))

kmer_text = sequences_df['seq'].apply(to_kmers)

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(kmer_text)

print("before:", sequences_df['seq'].iloc[0][:40])
print("after: ", X.toarray()[0][:40])
print("shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split #Debugging required, but this is used to split the data into training and testing sets. The training set is used to train the model, and the testing set is used to evaluate the model's performance.

X_train, X_test, y_train, y_test = train_test_split(
    X, test_size=0.2, random_state=42
)

print(X_train.shape[0], "to study")
print(X_test.shape[0], "to be tested on...")
```
