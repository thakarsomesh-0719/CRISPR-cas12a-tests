```python
from Bio import SeqIO
import pandas as pd

seq_file = open(r"C:\Users\thaka\OneDrive\Desktop\CRISPR-cas12a test\GCF_001274345.1_ViralProj293723_genomic.fna","rt")

sequences_obj = SeqIO.parse(seq_file, "fasta")
sequences = [seq for seq in sequences_obj]
sequences_df = pd.DataFrame([
    {"id": s.id, "seq": str(s.seq)} for s in sequences
])

print(sequences_df.columns.tolist())
print(f"Starting Base pairs {sequences[0].seq[:50]}")
```

```python
AT_content = sum(seq.seq.count('A') + seq.seq.count('T') for seq in sequences) / sum(len(seq.seq) for seq in sequences)
print(f"AT content of HPV sequence : {AT_content * 100:.3f}%")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

def to_kmers(seq, k=4):
    return ' '.join(seq[i:i+k] for i in range(len(seq) - k + 1))

kmer_text = sequences_df['seq'].apply(to_kmers)

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(kmer_text)

print("before:", sequences_df['seq'].iloc[0][:40])
print("after: ", X.toarray()[0][:40])
print("shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, test_size=0.2, random_state=42
)

print(X_train.shape[0], "to study")
print(X_test.shape[0], "to be tested on...")
```
