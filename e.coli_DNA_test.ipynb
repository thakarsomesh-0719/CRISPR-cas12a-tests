```python
import pandas as pd

url = ("https://archive.ics.uci.edu/ml/machine-learning-databases/"
       "molecular-biology/promoter-gene-sequences/promoters.data")

df = pd.read_csv(url, header=None, names=['label', 'name', 'sequence'])

sequences = df['sequence'].str.replace(r'\s', '', regex=True).str.lower()
labels = (df['label'].str.strip() == '+').astype(int)

print(len(sequences), "sequences")
print(sequences[0])
```

```python
from sklearn.feature_extraction.text import CountVectorizer

def to_kmers(seq, k=4):
    return ' '.join(seq[i:i+k] for i in range(len(seq) - k + 1))

kmer_text = sequences.apply(to_kmers)

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(kmer_text)

print("before:", sequences[0][:40])
print("after: ", X.toarray()[0][:40])
print("shape:", X.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42, stratify=labels
)

print(X_train.shape[0], "to study")
print(X_test.shape[0], "to be tested on...")
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
accuracy = model.score(X_test, y_test)
print(f"accuracy: {accuracy:.1%}")

i = 0
guess = model.predict(X_test[i])[0]
truth = y_test.iloc[i]

print("model says:", "promoter" if guess else "not a promoter")
print("actually:  ", "promoter" if truth else "not a promoter")
```
